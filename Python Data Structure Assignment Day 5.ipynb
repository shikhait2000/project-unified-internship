count the number of occurences of charachters in string named "sent" using dictionary and print the same.


```python

sent = "sent"

char_count = {}

for ch in sent:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1

print(char_count)
```
